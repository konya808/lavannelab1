```python
import math

# A simple program to calculate Ohm's Law
# This program will ask for voltage and resistance, then calculate current and power.

def calculate_ohms_law_simple():
    #  input from the user for voltage
    voltage_str = input("Please enter the voltage (in Volts): ")

    #  input from the user for resistance
    resistance_str = input("Please enter the resistance (in Ohms): ")

    try:
        # Convert input strings to numbers
        voltage = float(voltage_str)
        resistance = float(resistance_str)

        # Check if resistance is zero to avoid division error
        if resistance == 0:
            print("Error: Resistance cannot be zero. Please try again with a non-zero resistance.")
        else:
            # Calculate current using Ohm's Law (I = V / R)
            current = voltage / resistance

            # Calculate power (P = V * I)
            power = voltage * current

            # Display the results
            print(f"Calculated Current (Amperes): {current:.2f}")
            print(f"Calculated Power (Watts): {power:.2f}")

    except ValueError:
        # Handle cases where the user enters non-numeric input
        print("Invalid input! Please make sure to enter numbers for voltage and resistance.")
    except Exception as e:
        # Catch any other unexpected errors
        print(f"An unexpected problem occurred: {e}")

# Start the calculator
print("--- Simple Ohm's Law Calculator ---")
calculate_ohms_law_simple()
# The program finishes after one calculation.
```
