```python
import math

def calculate_ohms_law():
    while True:
        try:
            voltage_str = input("Enter voltage (v), or type 'q' to quit: ")
            if voltage_str.lower() == 'q':
                break
            voltage = float(voltage_str)

            resistance_str = input("Enter resistance (Ohms): ")
            resistance = float(resistance_str)

            if resistance != 0:
                current = voltage / resistance
                power = voltage * current
                print(f"Current (A): {current:.2f}")
                print(f"Power (W): {power:.2f}")
            else:
                print("Error: Resistance cannot be zero")
        except ValueError:
            print("Invalid input. Please enter numeric values for voltage and resistance.")
        except Exception as e:
            print(f"An unexpected error occurred: {e}")

        another_calculation = input("Do you want to perform another calculation? (yes/no): ")
        if another_calculation.lower() != 'yes':
            break

print("Ohm's Law Calculator")
calculate_ohms_law()
print("Thank you for using the Ohm's Law Calculator!")
```
